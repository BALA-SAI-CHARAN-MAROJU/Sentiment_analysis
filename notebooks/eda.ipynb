### 1. Ingest Dataset & Inspect Structure

```python
df = pd.read_csv(PROCESSED_DATA_FILE)
print(f'Total rows in processed corpus: {len(df)}')
df.head(5)
```

### 2. Class Balance Analysis

```python
class_counts = df['sentiment'].value_counts()
print(class_counts)

plt.figure(figsize=(6, 4))
colors = ['#2ea043', '#f85149', '#58a6ff']
class_counts.plot(kind='bar', color=colors, edgecolor='black')
plt.title('Class Distribution across Sentiment Categories', fontsize=12)
plt.xlabel('Sentiment')
plt.ylabel('Number of Samples')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

### 3. Text Preprocessing Demonstration
Notice how emojis are mapped to sentiment keywords rather than being dropped.

```python
raw_samples = [
    'Loving the brand new camera sensor on this phone! 😍✨ #tech @Google',
    'This is the WORST customer experience ever! 😡👎 http://complaint.com',
    'The quarterly update was uploaded today at 10 AM.'
]
for raw in raw_samples:
    print('RAW:', raw)
    print('CLEANED:', clean_text(raw))
    print('-' * 60)
```

### 4. Post Length Distribution

```python
df['char_len'] = df['text'].astype(str).str.len()
df['word_count'] = df['text'].astype(str).str.split().str.len()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df['char_len'].hist(ax=axes[0], bins=30, color='#484f58', edgecolor='white')
axes[0].set_title('Post Character Length Distribution')
axes[0].set_xlabel('Character Count')
axes[0].set_ylabel('Frequency')

df['word_count'].hist(ax=axes[1], bins=25, color='#58a6ff', edgecolor='black')
axes[1].set_title('Post Word Count Distribution')
axes[1].set_xlabel('Word Count')
plt.tight_layout()
plt.show()
```
